# Task 2 
Same sudoku image as Task 1.

Testing blockSize 5, 15 and 31 with the Mean method (keeping C fixed at 2), then
testing C values 2, 5 and 10 with Gaussian (keeping blockSize fixed at 15).

Changing one thing at a time so I can actually tell what each parameter does.

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("sudoku.png", cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Error: sudoku.png not found")
else:
    results = []
    titles = []

    # Test 3 blockSize values with Mean method
    for b in [5, 15, 31]:
        out = cv2.adaptiveThreshold(
            img, 255,
            cv2.ADAPTIVE_THRESH_MEAN_C,
            cv2.THRESH_BINARY,
            b, 2
        )
        results.append(out)
        titles.append("Mean, block=" + str(b) + ", C=2")

    # Test 3 C values with Gaussian method
    for c in [2, 5, 10]:
        out = cv2.adaptiveThreshold(
            img, 255,
            cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
            cv2.THRESH_BINARY,
            15, c
        )
        results.append(out)
        titles.append("Gaussian, block=15, C=" + str(c))

    plt.figure(figsize=(15, 9))

    plt.subplot(3, 3, 2)
    plt.imshow(img, cmap="gray")
    plt.title("Original")
    plt.axis("off")

    for i in range(6):
        plt.subplot(3, 3, i + 4)
        plt.imshow(results[i], cmap="gray")
        plt.title(titles[i])
        plt.axis("off")

    plt.tight_layout()
    plt.show()
```

### Analysis

**1. What happens when the neighbourhood is too small? (block=5)**

The patch is barely bigger than the thickness of a line, so the local average ends up
being calculated from the text itself. The characters break apart and the whole thing
gets really speckly and noisy.

**2. What happens when the neighbourhood is too large? (block=31)**

It starts acting like a global threshold again. The patch is big enough to cover both
the bright and dark parts of the page, so the "local" average isn't really local
anymore and the uneven lighting problem comes back.

**3. What happens when C is increased?**

C gets subtracted from the local average, so a bigger C means a lower threshold and
fewer white pixels. At C=10 the thin strokes start breaking up and some text gets
lost. At C=2 it's the opposite, there's more black noise around the letters.

**4. Which combination gives the cleanest text?**

Gaussian with blockSize 15 and C 5 looked best to me. blockSize around 11-15 with C
around 2-5 seems to be the good range.

**5. Which adaptive method works better on this image?**

Gaussian. It gives more weight to pixels closer to the centre of the block instead of
treating them all the same like Mean does, so the result comes out smoother with less
noise.
